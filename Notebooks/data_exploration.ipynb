```python
data_dictionary = []

for name, df in datasets.items():
    for col in df.columns:
        data_dictionary.append({
            "Dataset": name,
            "Column": col,
            "Data Type": str(df[col].dtype),
            "Missing Values": df[col].isnull().sum(),
            "Unique Values": df[col].nunique()
        })

dictionary_df = pd.DataFrame(data_dictionary)

dictionary_df.head()
```

```python
# Find common columns across datasets

column_map = {}

for name, df in datasets.items():
    for col in df.columns:
        if col not in column_map:
            column_map[col] = []
        column_map[col].append(name)

print("COMMON COLUMNS ACROSS DATASETS")
print("="*80)

for col, tables in column_map.items():
    if len(tables) > 1:
        print(f"\nColumn: {col}")
        print("Appears in:")
        for table in tables:
            print(f"   - {table}")
```

```python
#Finding unique coloumns or possible primary key columns
print("POTENTIAL PRIMARY KEYS")
print("="*80)

for name, df in datasets.items():

    print(f"\n{name}")

    for col in df.columns:

        unique_count = df[col].nunique()

        if unique_count == len(df):
            print(f"✓ {col} (Unique)")
```

```python
from pathlib import Path
import pandas as pd

# Base Folder
BASE_DIR = Path(r"C:\Users\P LIKITH VARMA\Downloads\bluestock capstone dataset")

# File Paths
fund_master_path = BASE_DIR / "01_fund_master.csv"
nav_history_path = BASE_DIR / "02_nav_history.csv"
```

```python
fund_master = pd.read_csv(fund_master_path)
nav_history = pd.read_csv(nav_history_path)

print("Fund Master Shape:", fund_master.shape)
print("NAV History Shape:", nav_history.shape)
```

```python
print("Fund Master Columns")
print(fund_master.columns.tolist())

print("\nNAV History Columns")
print(nav_history.columns.tolist())
```

```python
print("Number of Fund Houses:", fund_master["fund_house"].nunique())

fund_master["fund_house"].dropna().sort_values().unique()

print("Number of Categories:", fund_master["category"].nunique())

fund_master["category"].dropna().sort_values().unique()

print("Number of Sub-Categories:", fund_master["sub_category"].nunique())

fund_master["sub_category"].dropna().sort_values().unique()
```

```python
print("Number of Risk Categories:", fund_master["risk_category"].nunique())

fund_master["risk_category"].dropna().sort_values().unique()

amfi_codes = fund_master["amfi_code"].astype(str)

print("Total Records :", len(amfi_codes))
print("Unique Codes  :", amfi_codes.nunique())

amfi_codes.str.len().value_counts().sort_index()

fund_master["amfi_code"].head(20)
```

```python
fund_codes = set(
    fund_master["amfi_code"]
    .dropna()
    .astype(str)
)

nav_codes = set(
    nav_history["amfi_code"]
    .dropna()
    .astype(str)
)

missing_codes = fund_codes - nav_codes

print("Missing Codes:", len(missing_codes))

list(missing_codes)[:20]
```

```python
coverage = (
    (len(fund_codes) - len(missing_codes))
    / len(fund_codes)
) * 100

print(f"Coverage = {coverage:.2f}%")

print("="*50)
print("DATA QUALITY SUMMARY")
print("="*50)

print(f"Fund Houses      : {fund_master['fund_house'].nunique()}")
print(f"Categories       : {fund_master['category'].nunique()}")
print(f"Sub-Categories   : {fund_master['sub_category'].nunique()}")
print(f"Risk Categories  : {fund_master['risk_category'].nunique()}")

print("\nAMFI Validation")
print("-"*50)

print(f"Fund Master Codes : {len(fund_codes)}")
print(f"NAV History Codes : {len(nav_codes)}")
print(f"Missing Codes     : {len(missing_codes)}")
print(f"Coverage          : {coverage:.2f}%")

if len(missing_codes) == 0:
    print("""
Data Quality Summary
--------------------
• All AMFI scheme codes in fund_master are present in nav_history.
• No missing AMFI codes were detected.
• Referential integrity between the two datasets is maintained.
• Dataset is suitable for NAV trend analysis and fund-level analytics.
""")
else:
    print(f"""
Data Quality Summary
--------------------
• {len(missing_codes)} AMFI codes from fund_master are missing in nav_history.
• Coverage: {coverage:.2f}%
• Missing codes should be investigated before analysis.
""")
```
